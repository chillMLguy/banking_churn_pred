### Key Metric Analysis: F1-Score

While **Accuracy** provides a general sense of model performance, the **F1-Score** is the most important statistic in this context. 

* **Balance:** Our models show a discrepancy between Accuracy (high) and Recall/Precision (moderate).
* **Trade-off:** Accuracy can be misleading if the majority class is predicted correctly while the minority class is ignored. The F1-Score (the harmonic mean of Precision and Recall) ensures we are penalizing models that fail to capture the positive class effectively.
* **Winner:** Based on the F1-Score, **Random Forest** is our most robust model ($F1 \approx 0.618$), offering the best compromise between correctly identifying positives and minimizing false alarms.


### Recommendations & Future Work

* **Expand Hyperparameter Grid:** For better results, a larger search grid should be used. Specifically, increasing `n_estimators` for Random Forest.
* **Utilize Boosting Algorithms:** Future work should move beyond standard bagging (Random Forest) to boosting models like **XGBoost, LightGBM, or CatBoost**. These models iteratively correct the errors of previous trees and typically provide a significant jump in F1-Scores.
* **Feature Engineering:** Introducing more complex feature interactions or dimensionality reduction techniques may help in separating the classes more clearly.
